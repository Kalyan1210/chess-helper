```python
# Colab Cell 2 — engine wrapper
import chess
import chess.engine

# 1) launch Stockfish once
engine = chess.engine.SimpleEngine.popen_uci("/usr/games/stockfish")

def best_move_for(fen: str = None, depth: int = 15) -> chess.Move:
    """
    Return Stockfish’s best move for the given FEN.
    If fen is None, defaults to the standard starting position.
    """
    board = chess.Board(fen) if fen else chess.Board()
    info = engine.play(board, chess.engine.Limit(depth=depth))
    return info.move

# Example usage:
print("Starting-position best move:", best_move_for())
custom_fen = "r1bqkbnr/pppppppp/n7/8/8/8/PPPPPPPP/RNBQKBNR w KQkq - 0 1"
print("Custom FEN best move:", best_move_for(custom_fen, depth=10))

# When you’re completely done, shut down the engine:
engine.quit()
```
